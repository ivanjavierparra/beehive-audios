> **OSBH: Open Source Beehive Project**
- Proyecto de ciencia ciudadana global con registros heterogéneos de múltiples países (Norteamérica, Australia, Europa). Sin especificaciones de hardware comunes. Porción pública disponible en Zenodo. 
- Acceso: https://zenodo.org/records/321345

- no lo usamos. 
- **Es la union de dos datasets: NU-Hive y OSBH.**
- Nolasco, I., & Benetos, E. (2018). To bee or not to bee: An annotated dataset for beehive sound recognition [Dataset]. 
- Zenodo. https://doi.org/10.5281/zenodo.1321278


> **Beehive buzz anomalies [yevheniiklymenko]**
- El conjunto de datos de " To bee or not to bee" se dividió en archivos wav de 2 segundos, se escaló a 44,1 kHz y se clasificó en estados "normal", "no es una abeja" y "no hay reina".
- https://www.kaggle.com/datasets/yevheniiklymenko/beehive-buzz-anomalies

> **BeeHive Audio Dataset with Queen and without Queen [harshkumar1711]**
- Este conjunto de datos está diseñado para respaldar la investigación en aprendizaje automático y aprendizaje profundo, centrada en la detección de la presencia o ausencia de una abeja reina en una colmena mediante señales de audio y espectrogramas. Puede utilizarse para desarrollar modelos de detección, clasificación y monitorización bioacústica de eventos acústicos.
